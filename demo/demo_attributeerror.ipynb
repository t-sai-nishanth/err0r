# BookError Demo 2 — Out-of-Order AttributeError Tracing

This notebook demonstrates how **BookError** catches out-of-order variable shadowing/overwrites.
1. **Cell 1** initializes a model instance `model = LinearRegression()`.
2. **Cell 2** reassigns `model = 'scikit-learn'` (overwriting object with string).
3. **Cell 3** calls `model.predict(X)`, raising an `AttributeError`.

```python
%load_ext bookerrror
from sklearn.linear_model import LinearRegression
```

```python
# Cell 2: Initialize model
model = LinearRegression()
```

```python
# Cell 3: Reassign variable 'model' to a string (out of order overwrite)
model = 'scikit-learn'
```

```python
# Cell 4: Attempt to call .predict()
# BookError traces back to Cell 3's variable overwrite!
model.predict([[1, 2]])
```
