# BookError Demo 1 — Out-of-Order KeyError Tracing

This notebook demonstrates how **BookError** pinpoints the root cause of an out-of-order execution error.
In this scenario:
1. **Cell 2** loads a dataset into variable `df`.
2. **Cell 3** drops the `'age'` column from `df`.
3. **Cell 4** (or an earlier cell re-run afterward) attempts to access `df[['age', 'income']]`, causing a `KeyError`.

```python
%load_ext bookerrror
import pandas as pd
```

```python
# Cell 2: Load data
df = pd.read_csv('data.csv')
df.head()
```

```python
# Cell 3: Drop column 'age'
df = df.drop(columns=['age'])
```

```python
# Cell 4: Access dropped column 'age'
# Running this cell triggers BookError's Scikit-Learn style pipeline visualizer!
result = df[['age', 'income']]
```
